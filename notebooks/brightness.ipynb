```python
import cv2
import os
import numpy as np

# --- Configuration ---
input_folder = "D:/CHARGER/Abumentation/Charger_v1.0/charger_v0.8/images"      # Replace with your image folder path
output_folder = "D:/CHARGER/Abumentation/Charger_v1.0/charger_v0.8/images/brightened_images"       # Output folder
                 # Increase value (try 30–100)

# --- Make output folder if not exists ---
os.makedirs(output_folder, exist_ok=True)

# --- Process each image ---
for filename in os.listdir(input_folder):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp')):
        img_path = os.path.join(input_folder, filename)
        img = cv2.imread(img_path)

        if img is None:
            print(f"Failed to load: {filename}")
            continue

        # Brighten image
        bright_img = cv2.convertScaleAbs(img, alpha=1, beta=50)

        # Save brightened image
        output_path = os.path.join(output_folder, filename)
        cv2.imwrite(output_path, bright_img)
        print(f"Saved: {output_path}")
```
