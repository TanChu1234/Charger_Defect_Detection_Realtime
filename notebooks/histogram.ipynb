```python
import cv2
import matplotlib.pyplot as plt

# Đọc ảnh và chuyển sang ảnh xám
img_path1 = r"C:\Users\Engineer\Pictures\1620HV.bmp"
img_path2 = r"C:\Users\Engineer\Pictures\LDV16.bmp"
img_path3 = r"C:\Users\Engineer\Pictures\LLD15.bmp"

img1 = cv2.imread(img_path1)
img2 = cv2.imread(img_path2)
img3 = cv2.imread(img_path3)

gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
gray3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)

# Tính và normalize histogram
hist1 = cv2.calcHist([gray1], [0], None, [256], [0, 256])
hist2 = cv2.calcHist([gray2], [0], None, [256], [0, 256])
hist3 = cv2.calcHist([gray3], [0], None, [256], [0, 256])

hist1 = cv2.normalize(hist1, hist1, 0, 1, cv2.NORM_MINMAX)
hist2 = cv2.normalize(hist2, hist2, 0, 1, cv2.NORM_MINMAX)
hist3 = cv2.normalize(hist3, hist3, 0, 1, cv2.NORM_MINMAX)

# Vẽ 3 histogram trên cùng 1 biểu đồ
plt.figure(figsize=(8, 5))
plt.plot(hist1, label='1620HV', color='red')
plt.plot(hist2, label='LDV16', color='green')
plt.plot(hist3, label='LLD15', color='blue')
plt.title("Normalized Histograms of 3 Images")
plt.xlabel("Pixel Value")
plt.ylabel("Normalized Frequency")
plt.legend()
plt.grid()
# plt.plot(hist1, label='Image 1', color='red')
# plt.show()
# plt.plot(hist2, label='Image 2', color='green')
# plt.show()
# plt.plot(hist3, label='Image 3', color='blue')
# plt.show()
```
